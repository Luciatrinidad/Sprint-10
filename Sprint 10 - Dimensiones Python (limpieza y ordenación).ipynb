### Unificamos las tablas

```python
tablacompleta= pd.merge(dim_macroeconomica, dim_financiera, on=["Country", "Year"], how="outer")
tablacompleta= pd.merge(tablacompleta, dim_rentas_extract, on=["Country", "Year"], how="outer")
tablacompleta= pd.merge(tablacompleta, dim_prod_extract, on=["Country", "Year"], how="outer")
tablacompleta= pd.merge(tablacompleta, dim_demo_migra, on=["Country", "Year"], how="outer")
tablacompleta= pd.merge(tablacompleta, dim_ambiental, on=["Country", "Year"], how="outer")


tablacompleta
```

### Latinoamericanizamos la muestra y recortamos el periódo a partir de 1990

```python
Muestra_Latam=tablacompleta[tablacompleta['Country'].isin(['Argentina', 'Bolivia', 'Brazil','Chile','Colombia', 'Costa Rica','Dominican Republic','Ecuador', 'El Salvador','Guatemala', 'Honduras', 'Mexico', 'Panama','Paraguay', 'Peru','Uruguay', 
                         'Venezuela, RB'  ])]
Muestra_Latam['Year']=Muestra_Latam["Year"].astype(int)
Muestra_Latam=Muestra_Latam[Muestra_Latam["Year"] >= 1990]
Muestra_Latam.reset_index()
```

## Test Estadístico (no normal)

```python
import pandas as pd
from scipy.stats import shapiro


columnas_numericas = Muestra_Latam.select_dtypes(include='number').columns
columnas_numericas = [col for col in columnas_numericas if col.lower() != 'year']  # excluye 'Year' sin importar mayúsculas

# Diccionario para guardar resultados
resultados = {}

for col in columnas_numericas:
    data = Muestra_Latam[col].dropna()
    if 3 <= len(data) <= 5000:
        stat, p = shapiro(data)
        resultados[col] = {'stat': stat, 'p': p, 'normal': p > 0.05}
    else:
        resultados[col] = {'stat': None, 'p': None, 'normal': None, 'msg': 'Demasiados datos o insuficientes para Shapiro'}

# Convertir los resultados en un DataFrame
resultados_df = pd.DataFrame.from_dict(resultados, orient='index')
resultados_df

```

```python
estadistica= resultados_df['stat']
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=estadistica)

```
